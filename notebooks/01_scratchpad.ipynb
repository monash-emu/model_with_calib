```python
import pandas as pd
```

```python
observations = {
    1: 50,
    3: 500,
}

obs_times = [1, 3]
cases = [50, 100]

init_state = {
    "sus": 999,
    "inf": 1,
    "rem": 0,
}

# Proposed transitions
trans_df = pd.DataFrame(
    {
        "inc": [300, 400],
        "rec": [1, 20],
    }, 
    index=observations.keys(),
)
```

```python
# Map from transitions and initial conditions to states
state_times = [0] + list(observations.keys())
states_df = pd.DataFrame(columns=init_state.keys(), index=state_times)
states_df.loc[0, :] = init_state

def update(sus, inf, rem, inc, rec):
    sus -= inc
    inf += inc - rec
    rem += rec
    return sus, inf, rem

sus, inf, rem = init_state.values()
for t in observations.keys():
    sus, inf, rem = update(sus, inf, rem, trans_df.loc[t, "inc"], trans_df.loc[t, "rec"])
    states_df.loc[t, :] = sus, inf, rem
```
