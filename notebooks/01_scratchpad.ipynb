- Calibration integrated with model inference (whatever that means)
- Calibration not necessarily sequential
- Explicit, integer-valued states
- Integer-valued states, stochastic transitions

```python
import numpy as np
import pandas as pd
```

States: sus, inf, rem

Transitions: inc, rec

```python
pop = 1000
seed = 1
obs_times = [1, 3]
cases = [50, 100]

transitions = {
    "inf": [300, 400],
    "rec": [1, 20],
}
trans_df = pd.DataFrame(transitions, index=obs_times)

init_state = {
    "sus": pop,
    "inf": seed,
    "rem": 0,
}
state_times = [0] + obs_times
states_df = pd.DataFrame(columns=init_state.keys(), index=state_times)
states_df.loc[0, :] = init_state

def update(sus, inf, rem, inc, rec):
    sus -= inc
    inf += inc - rec
    rem += rec
    return sus, inf, rem

sus, inf, rem = states_df.loc[0, :]
for t in obs_times:
    new_states = update(sus, inf, rem, trans_df.loc[t, "inf"], trans_df.loc[t, "rec"])
    states_df.loc[t, :] = new_states
```
